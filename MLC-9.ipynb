```python
# Import required libraries
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report

# Create email dataset
data = {
    "Email": [
        "Win a free lottery prize now",
        "Claim your free cash reward",
        "Congratulations you won a gift",
        "Get a free mobile phone today",
        "Click here to claim your prize",
        "You have won a lottery",
        "Earn money quickly from home",
        "Limited offer buy now",
        "Free gift waiting for you",
        "Urgent claim your reward",
        "Meeting scheduled for tomorrow",
        "Please submit your assignment",
        "Your exam timetable is ready",
        "Let us meet for lunch",
        "Project discussion at 10 AM",
        "Please check your email",
        "Your class starts at 9 AM",
        "Thank you for your response",
        "Please attend the meeting",
        "Your report has been submitted"
    ],

    "Category": [
        "Spam", "Spam", "Spam", "Spam", "Spam",
        "Spam", "Spam", "Spam", "Spam", "Spam",
        "Not Spam", "Not Spam", "Not Spam",
        "Not Spam", "Not Spam", "Not Spam",
        "Not Spam", "Not Spam", "Not Spam",
        "Not Spam"
    ]
}

df = pd.DataFrame(data)

# Separate features and labels
X = df["Email"]
y = df["Category"]

# Convert text into numerical features
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(X)

# Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

# Train Naive Bayes model
model = MultinomialNB()
model.fit(X_train, y_train)

# Predict email categories
y_pred = model.predict(X_test)

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

print("Email Classification Results:")
print("Predicted Categories:", y_pred)
print("Actual Categories:", list(y_test))

print("\nAccuracy:", accuracy * 100, "%")

# Classification report
print("\nClassification Report:")
print(classification_report(y_test, y_pred))

# Test a new email
new_email = ["Congratulations! Claim your free prize now"]

new_email_vector = vectorizer.transform(new_email)

prediction = model.predict(new_email_vector)

print("\nNew Email:", new_email[0])
print("Email Category:", prediction[0])
```
